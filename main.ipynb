```python
from rasterizer import lineRasterizer

def draw_line_dda(fb, x1, y1, x2, y2, color=(0, 0, 0)):
    y = float(y1)
    dy = (y2 - y1) / (x2 - x1)

    for x in range(x1, x2 + 1):
        py = int(round(y))
        fb[x, py] = color
        y += dy

p1 = (1,11)
p2 = (11, 1)

lineRasterizer(draw_line_dda, (p1, p2))
```
